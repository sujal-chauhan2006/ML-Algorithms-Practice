```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
x,y = make_regression(n_samples = 4,n_features=1,n_informative=1,noise=80,random_state =13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
```

```python
# Apply OLS
reg = LinearRegression()
reg.fit(x,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red')
```

```python
# Applying Gredinet Decent assuming m = 78.35
# And assuming that the starting value for intercept is b = 0
y_pred = ((78.35 * x) + 0).reshape(4)
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label = 'OLS')
plt.plot(x,y_pred,color='blue',label = 'b = 0')
plt.legend()
plt.show()
```

```python
m = 78.35
b = 0

loss_slope = -2 * np.sum(y - m*x.ravel() - b)
loss_slope
```

```python
# take learning rate = 0.1

lr = 0.1

step_size = loss_slope * lr
step_size
```

```python
# New Intercept
b = b - step_size
b
```

```python
y_pred1 = ((78.35 * x) + b).reshape(4)
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label = 'OLS')
plt.plot(x,y_pred,color='blue',label = 'b = 0')
plt.plot(x,y_pred1,color='green',label = 'b = 62.78')
plt.legend()
plt.show()
```

```python
b = -100
m = 78.35
lr = 0.1

epochs = 10

for i in range(epochs):
    loss_slope = -2 * np.sum(y - m*x.ravel() - b)
    b = b -(lr * loss_slope)

    y_pred = m * x + b
    plt.plot(x,y_pred)

plt.scatter(x,y)
```
